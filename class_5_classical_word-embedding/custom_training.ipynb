```python
folder_path = r"C:\Users\ASUS\Desktop\Gen AI Full stak\class_5_classical_word-embedding\data"
story = []
for filename in os.listdir(folder_path):
  file_path=os.path.join(folder_path,filename)
  with open(file_path,encoding='unicode_escape') as f: #encoding='unicode_escape' tells Python how to read the bytes in the file and convert them into text 
    corpus=f.read()                                     # Hello\nWorld with encoding='unicode_escape  we can read as Hello and then in new line world simillary it deals with Caf\u00e9 or any special tings 
  raw_sent=sent_tokenize(corpus)
  for sent in raw_sent:
    story.append(simple_preprocess(sent))
```

```python
custom_model =gensim.models.Word2Vec(story, vector_size=150, window=5, min_count=1, workers=4, sg=1) # min_count=1 means that we will consider all the words in the corpus, even if they appear only once. If we set min_count=2, then we will ignore all the words that appear only once in the corpus.
                                                                               # vector_size=100 means that we will create a 100-dimensional vector for each word in the corpus. If we set vector_size=50, then we will create a 50-dimensional vector for each word in the corpus.
                                                                                # sg=1 means that we will use the skip-gram model to train the word2vec model. If we set sg=0, then we will use the CBOW model to train the word2vec model.
     

```

```python
custom_model.build_vocab(story)
```

```python
custom_model.corpus_count # it means the total number of sentences
```

```python
custom_model.train(story, total_examples=custom_model.corpus_count, epochs=30)
```

```python
from gensim.utils import simple_preprocess

story = []

max_lines = 50_000

for row in dataset["train"]:
    text = row["text"].strip()

    if not text:
        continue

    tokens = simple_preprocess(text)

    if tokens:
        story.append(tokens)

    if len(story) >= max_lines:
        break

print("Sentences:", len(story))
```

```python
import logging
from gensim.models import Word2Vec
from gensim.models.callbacks import CallbackAny2Vec
from tqdm.auto import tqdm

logging.getLogger("gensim").setLevel(logging.ERROR)


class ProgressBar(CallbackAny2Vec):
    def __init__(self, epochs):
        self.pbar = tqdm(
            total=epochs,
            desc="Training Word2Vec"
        )

    def on_epoch_end(self, model):
        self.pbar.update(1)

    def on_train_end(self, model):
        self.pbar.close()


epochs = 5

model = Word2Vec(
    sentences=story,
    vector_size=100,
    window=5,
    min_count=2,
    workers=8,
    sg=0,              # CBOW is faster than Skip-gram
    epochs=epochs,
    callbacks=[ProgressBar(epochs)]
)
```
