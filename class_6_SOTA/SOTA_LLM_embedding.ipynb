Semantic(similarity search) search

To calculate the similarty 
1. Dot product
2. Eculidiean Distance 
3.Cosine Similarity


```python
def dot_product(a, b):
    return np.dot(a, b)
```

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
def euclidean_distance(a, b):
    a = np.array(a)
    b = np.array(b)
    return np.linalg.norm(a - b)
```

```python
print("=== Cosine Similarity (Higher is better) ===")
print("Range: -1 to 1")

print("\n=== Dot Product (Higher is better) ===")
print("Range: -∞ to +∞ (depends on vector magnitude)")

print("\n=== Euclidean Distance (Lower is better) ===")
print("Range: 0 to +∞")
```

Lets check the similart search using this searches

```python
query = "How to reduce heart disease risk?"
```

```python
documents = [
    "Eating fiber reduces heart risk.",
    "Fruits and vegetables lower cardiovascular disease chances.",
    "Buying a new car improves driving comfort.",
    "Regular exercise improves heart health."
]
```

```python
query_embeddings = goole_embedding_model.embed_query(query)
```

```python
doc_embeddings = goole_embedding_model.embed_documents(documents)
```

```python
len(doc_embeddings)
```

```python
len(query_embeddings)
```

```python
results=[]
```

```python
import numpy as np
```

```python
results = []

for doc, doc_emb in zip(documents, doc_embeddings):
    
    results.append({
        "query": query,
        "document": doc,
        "cosine": cosine_similarity(query_embeddings, doc_emb),
        "dot_product": dot_product(query_embeddings, doc_emb),
        "euclidean": euclidean_distance(query_embeddings, doc_emb)
        
    })
```
