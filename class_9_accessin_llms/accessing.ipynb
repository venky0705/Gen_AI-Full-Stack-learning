# huggingface
text to text: free

```python
import os
HUGGINGFACEHUB_API_TOKEN = os.getenv ("HUGGINGFACEHUB_API_TOKEN")  # This will return None if the environment variable is not set
if HUGGINGFACEHUB_API_TOKEN:
    os.environ["HUGGINGFACEHUB_API_TOKEN"] = HUGGINGFACEHUB_API_TOKEN
```

```python
from langchain_huggingface import HuggingFaceEndpoint
```

```python
llm = HuggingFaceEndpoint(
    #repo_id="deepseek-ai/DeepSeek-R1-0528",
    repo_id="deepseek-ai/DeepSeek-V4-Pro",
    task="text-generation",
    max_new_tokens=512
)
```

```python
from langchain_huggingface import ChatHuggingFace

chat_model = ChatHuggingFace(llm=llm)
```

Prompt(Input message) ---->LLM----->output(output message or AI-Message)

user_message

or

system_message user_message

we have two way of defining ths system and user message

first simple tuple format(older way of passing message but still relevant)
messages = [ ("system", "You are a helpful assistant that translates English to Hindi. Translate the user sentence.",), ("human", "I love programming.") ]

second using SystemMessage class and HumanMessage class(latest way of defining message) latest LangChain variant messages = [ SystemMessage(content="You're a helpful assistant"), HumanMessage( content="What happens when an unstoppable force meets an immovable object?"), ]
SystemMessage or system -->we are defining the permanent behviour of llm in current active session HumanMessage or human --> whatever human wanted to provide input it is that message

SystemMessage just to set the role or the behaviour of the LLM for the active session.



```python
from langchain_core.messages import SystemMessage, HumanMessage
messages = [
    SystemMessage(content="You're a helpful assistant"),
    HumanMessage(content="What happens when an unstoppable force meets an immovable object?")
]
```

```python
ai_msg = chat_model.invoke(messages)
```

```python
print(ai_msg.content)
```

# google gemini

```python
import base64
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage

# 1. Image ko base64 me convert karo
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")

base64_image = encode_image(
    r"C:\Users\ASUS\Pictures\Screenshots\Screenshot 2026-08-09 153357.png"
)
```

```python
# 2. Gemini model initialize karo
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=GOOGLE_API_KEY
)
```

```python
# 3. Text + image message banao
message = HumanMessage(
    content=[
        {
            "type": "text",
            "text": "Describe this image in detail."
        },
        {
            "type": "image_url",
            "image_url": f"data:image/png;base64,{base64_image}"
        }
    ]
)
```

```python
# 4. Model invoke karo
response = llm.invoke([message])
```

```python
from IPython.display import Markdown, display

response = llm.invoke([message])

text = response.content[0]["text"]

display(Markdown(text))
```
