# google text to speech

```python
from google import genai
from google.genai import types
import wave

# Use your Gemini API key
client = genai.Client(
    api_key="paste the key"
)

def save_wav(filename, pcm_data):
    with wave.open(filename, "wb") as wf:
        wf.setnchannels(1)
        wf.setsampwidth(2)
        wf.setframerate(24000)
        wf.writeframes(pcm_data)

response = client.models.generate_content(
    model="gemini-3.1-flash-tts-preview",
    contents="Say cheerfully: Welcome to my Generative AI course!",
    config=types.GenerateContentConfig(
        response_modalities=["AUDIO"],
        speech_config=types.SpeechConfig(
            voice_config=types.VoiceConfig(
                prebuilt_voice_config=types.PrebuiltVoiceConfig(
                    voice_name="Kore"
                )
            )
        )
    )
)

audio_data = (
    response
    .candidates[0]
    .content
    .parts[0]
    .inline_data
    .data
)

save_wav("output.wav", audio_data)

print("Audio generated successfully: output.wav")
```

```python
from google import genai
from google.genai import types
from IPython.display import Audio, display
import wave

API_KEY = "past the key"

client = genai.Client(api_key=API_KEY)


def text_to_speech(text, filename="speech.wav"):

    response = client.models.generate_content(
        model="gemini-3.1-flash-tts-preview",
        contents=text,
        config=types.GenerateContentConfig(
            response_modalities=["AUDIO"],
            speech_config=types.SpeechConfig(
                voice_config=types.VoiceConfig(
                    prebuilt_voice_config=types.PrebuiltVoiceConfig(
                        voice_name="Kore"
                    )
                )
            )
        )
    )

    audio_data = (
        response
        .candidates[0]
        .content
        .parts[0]
        .inline_data
        .data
    )

    with wave.open(filename, "wb") as wf:
        wf.setnchannels(1)
        wf.setsampwidth(2)
        wf.setframerate(24000)
        wf.writeframes(audio_data)

    print("Audio created:", filename)

    display(Audio(filename))
```

# speach to text

```python
import sounddevice as sd
from scipy.io.wavfile import write
from google import genai

client = genai.Client(
    api_key="past the key"
)

def record_and_transcribe(seconds=5):
    sample_rate = 16000
    filename = "my_voice.wav"

    print(f"Speak now for {seconds} seconds...")

    recording = sd.rec(
        int(seconds * sample_rate),
        samplerate=sample_rate,
        channels=1,
        dtype="int16"
    )

    sd.wait()

    write(
        filename,
        sample_rate,
        recording
    )

    print("Recording finished.")
    print("Uploading audio...")

    audio_file = client.files.upload(
        file=filename
    )

    print("Transcribing...")

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=[
            "Transcribe exactly what is spoken. Return only the text.",
            audio_file
        ]
    )

    print("\nTranscription:")
    print(response.text)

    return response.text
```

```python
text = record_and_transcribe(seconds=5)
```
